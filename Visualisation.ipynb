```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create sample data using NumPy and Pandas
data = {
    'study_hours': [1, 2, 2.5, 3, 4, 4.5, 5, 6, 7, 8],
    'exam_score': [55, 60, 62, 68, 75, 74, 80, 88, 91, 95]
}
df = pd.DataFrame(data)

# Print the DataFrame to see your data
print(df)

# 2. Create a beautiful plot using Seaborn and Matplotlib
plt.figure(figsize=(7, 6)) # Use Matplotlib to set the figure size
sns.scatterplot(x='study_hours', y='exam_score', data=df, s=100) # Use Seaborn for the plot

# Add titles and labels for clarity
plt.title('Study Hours vs. Exam Score')
plt.xlabel('Hours Spent Studying')
plt.ylabel('Exam Score (out of 100)')
plt.grid(True) # Add a grid for better readability

# 3. Show the plot
plt.show()


plt.plot(df)
```

```python
#basic  plotting with seaborn
tips=sns.load_dataset('tips')
tips
```

```python
#create a scatter plot 
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("scatter plot of total bill vs tip")
plt.show()
```

```python
#line plot 
sns.lineplot(x='size',y='total_bill',data= tips)
plt.title("line plot of total bill")
plt.show()
```

```python
#categorical plots 
##bar plot 
sns.barplot(x='day',y='total_bill',data=tips)
plt.title('bar plot ')
plt.show
```

```python
#box plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
#violin plot
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
#histograms
sns.histplot(tips['total_bill'],bins=10,kde=True)
#kde =True was responsible for the curve that smoothens the graph
```

```python
#KDE plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
#pair plot 
sns.pairplot(tips)
```

```python
#correlation heatmap

corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```
