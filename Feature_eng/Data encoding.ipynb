### Nominal OHE Encoding

one hot encoding also know as nominal encoding is a technique used to represent categorical data as numerical data which is more suitable for machine learning algorithms in this technique each category is represented as a binary vector where each bit corresponds to a unique category.For example,if we have a categorical variable "color" with three possible values(red,green,blue) we can represent one hot encoding as follows


1. Red:[1,0,0]
2. Green:[0,1,0]
3. Blue:[0,0,1]


Sparse matrix : u should not use ohe when u have many many categories 

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
#create a simple dataframe

df=pd.DataFrame({
    'color':['red','blue','green','red','blue']
})
```

```python
df.head()
```

```python
#create isntance of one hot encoding
encoder =OneHotEncoder()
```

```python
#perform fit and transform
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
#why in the output red went 0,0,1 as the function did alphabetical shorting 
#it led to red to hold the last spot and blue to hold the first
```

```python
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

### label encoding

label encoding and ordinal encoding are two techniques used to encode categorical data as numerical data 

label encoding involves assigning  a unique numerical label to each category in the variable .the label are usually assigned in alphabetical order or based on the frequency of the categories. for example if we have a categorical variable 'color' with three possible values


1. Red:1
2. Green:2
3. Blue:3

### Ordinal encoding 

converts categorical values into ordered integers based on a defined or natural ranking.
It assigns each category a number like 0, 1, 2, 3… depending on the order.
Useful when categories have a meaningful order (e.g., low < medium < high).
Not suitable for unordered categories because it introduces false numerical relationships.

1. HighSchool:1
2. college :2
3. Graduate:3
4. post-graduate:4


```python
import pandas as pd 
from sklearn.preprocessing import OrdinalEncoder
```

```python
df1 = pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
df1
```

```python
#create an instance of ordinalencoder and the fit_transform

encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df1[['size']])
```

```python
encoder.fit_transform([['small']])
```

### Target guided ordinal encoding
Target-guided ordinal encoding is a technique where categorical values are converted into ordered integers based on the target variable (y) instead of natural order.

Here’s how it works (simple):

1. For each category, compute a target statistic (usually mean of the target, e.g., probability of class 1).

2. Sort categories based on this value.

3. Assign integers (0,1,2…) according to the order.

```python
import pandas as pd 

#create a simple dataframe

df2 = pd.DataFrame({
    'city':['New york','Mumbai','shanghai','beijing','london','singapore'],
    'Price':[250,167,189,494,204,320]
})
```

```python
df2
```

```python
'''we will consider price as target variable and 
try to convert city(categorical variabel )on the basis of target variable
'''
mean_price=df2.groupby('city')['Price'].mean().to_dict()
```

```python
mean_price
```

```python
df2['city_encoded']=df2['city'].map(mean_price)
```

```python
df2[['Price','city_encoded']]
```
