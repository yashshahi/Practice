## Minimum,Maximum,median,Q1,Q3,IQR

```python
import numpy as np

```

```python
lst_marks = [45,76,87,43,78,97,34,65,76,99,32,54,67,74]
minimum,Q1,median,Q3,maximum=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*(IQR)
higher_fence=Q3+1.5*(IQR)
```

```python
higher_fence #anything greater than this will be outlier
```

```python
lower_fence #anything that is less than this will be consider as outlier
```

```python
import seaborn as sns
sns.boxplot(lst_marks)
#if there is a outlier the larger or the lower values will be shown as dot
```
