### Handling imbalanced dataset

Imbalance dataset will create a bias which will make it harder to model using ML

1. Up sampling
2. Down sampling

```python
import numpy as np 
import pandas as pd 

#set the random seed reproducibility
np.random.seed(123)

#create a dataframe with two classes
n_samples=1000
class_0_ratio=0.9 #multiplying this by samples (which will produce 900 data points)
n_class_0 = int(n_samples *class_0_ratio)
n_class_1 = n_samples-n_class_0# and remaining 100 will produced here
```

```python
n_class_0,n_class_1
```

```python
#create my df with imbalance dataset

class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df['target'].value_counts() #checks the class distribution
```

### upsampling
It works by increasing the number of samples in the minority class, usually by duplicating rows or generating synthetic data (e.g., SMOTE).
This helps machine learning models learn equally from all classes.

```python
### upsampling

df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
#the resample will try to extrapolate the minority 
```

```python
df_minority_upsampled=resample(df_minority,replace=True,#sample with replacement
        n_samples=len(df_majority),   
        random_state=42 #fixes seed
)
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled=pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

### when to upsample

You should upsample when:

1. Your dataset is imbalanced and the minority class has very few samples.

2. You want the model to learn enough from the minority class to make accurate predictions.

3. You prefer to increase the dataset size (by duplicating or synthesizing samples) rather than remove majority-class data.


### Downsampling
Downsampling is a technique used to fix imbalanced datasets by reducing the number of samples in the majority class.
It does this by randomly removing rows from the overrepresented class.
This helps the model avoid bias toward the majority class.

```python
import numpy as np 
import pandas as pd 

#set the random seed reproducibility
np.random.seed(123)

#create a dataframe with two classes
n_samples=1000
class_0_ratio=0.9 #multiplying this by samples (which will produce 900 data points)
n_class_0 = int(n_samples *class_0_ratio)
n_class_1 = n_samples-n_class_0# and remaining 100 will produced here

#create my df with imbalance dataset

class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=2,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=2,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})

df1=pd.concat([class_0,class_1]).reset_index(drop=True)

df1['target'].value_counts() #checks the class distribution
```

```python
### downsampling

df_minority=df[df1['target']==1]
df_majority=df[df1['target']==0]
```

```python
df_majority_downsampled=resample(df_majority,replace=False,#replace is false as we have to reduce the data points
        n_samples=len(df_minority),   
        random_state=42 #fixes seed
)
```

```python
df_majority_downsampled.shape
```

```python
df_majority_downsampled.head
```

```python
df_downsampled=pd.concat([df_minority,df_majority_downsampled])
```

```python
df_downsampled['target'].value_counts()
```

### When to Downsample 

You should downsample when:

1. Your dataset is highly imbalanced, and the majority class has far more samples than the minority class.

2. You want to reduce bias in your model toward the majority class.

3. You prefer a smaller, faster-to-train dataset, and removing some majority samples won’t harm important patterns.
