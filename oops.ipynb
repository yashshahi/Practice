## Inheritance in Python

Inheritance is a fundamental concept in Object-Oriented Programming(OOP) that allows a class to inherit attributes and methods from another class.

```python
#Inheritance
#parent class

#singel inheritance because one child class is inhheriting parent class

class car:
    def __init__ (self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype= enginetype

    def drive(self):
        print(f'the person will drive the {self.enginetype} car')

```

```python
car1=car(4,5,"petrol")
car1.drive()
```

```python
class Tesla(car):
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype) #we  are calling the init method from the parent class
        self.is_selfdriving=is_selfdriving
        
    def selfdriving(self):
        print(f"tesla supports:{self.is_selfdriving}")

```

```python
tesla1=Tesla(4,4,"electric",True)
tesla1.selfdriving()
```

```python
#Multiple inheritance
#when a class inherits more than one base class

class Animal:#base class 1
    def __init__(self,name):
        self.name=name
    def speak(self):
        print("subclass must implement this method ")


#base class 2 
class pet:
    def __init__(self,owner):
        self.owner=owner
    
#derived class
class Dog(Animal,pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        pet.__init__(self,owner)

    def speak(self):
        return f"{self.name} says woof"

##create an obj

dog= Dog("Buddy","yash")
print(dog.speak())
print(f"owner:{dog.owner}")

```

### Polymorphism

polymorhism is a core concept in OOP that allows object of different classes to be treated as obj of common superclass, it provides ways to perform a single action in different form ,Polymorphism is typically achieved through method overriding and interfaces

### Method overriding 

method overriding allows a child class to provide a specific implementation of a method that is already defined in its parent class

```python
##base class

class Animal:
    def speak(self):
        return "sound of the animal"

#derived class 1 

class Dog(Animal):
    def speak(self):
        return "woof!" #child class has provided its own specific implementation

#derived class 2
class cat(Animal):
    def speak(self):
        return "Meow!!"

##function that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())


dog = Dog()
cat = cat()
print(dog.speak())
print(cat.speak())
animal_speak(dog)
```

```python
#polmorphism with functions and methods 

#base class
class Shape:
    def area(self):
        return "the area of the figure"

#derived class 1 

class rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width*self.height

##derived class 2
class circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius**2

##function that demonstrate polymorphism
def print_area(shape):
    print(f'the area is {shape.area()}')


rectangle = rectangle(4,5)
circle = circle(3)

print_area(rectangle)
print_area(circle)
```

## polymorphism with abstract base class

abstract base class (ABC) are used to define common methods for a group of related objects .They can enforce that derived classes implement particular methods, promoting consistency across different implementations

```python
from abc import ABC,abstractmethod

##define an abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

#derived class
class Car(Vehicle):
    def start_engine(self):
        return "car engine started"

#derived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "motorcycle engine started "

#function that demonstrates polymorphism
def start_Vehicle(Vehicle):
    print(Vehicle.start_engine())


## we create obj of car and motorcycles

car = Car()
motorcycle =Motorcycle()


start_Vehicle(car)
```

### Encapsulation and Abstraction 

Encapsulation and abstraction are two fundamental principles of OOPS that help in designing robust,maintainable and reusable code. Encapsulation involves bundling data and methods that operate on data within a single unit ,while abstraction involves hiding complex implementation details and exposing only the necessary features

### Encapsulation

encapsulation is the concept of wrapping data(variables) and methods together as a single unit. it restricts direct access to some of the objects components which is a means of preventing accidental interference and misuse of data

```python
### Encapsulatiokn with getter and setter methd
#Public , protected and private variables

class Person:
    def __init__(self,name,age):
        self.name=name   #public variables
        self.age=age     #public variables

def get_name(person):
    return person.name


person = Person("Krish",34)
get_name(person) 
'''
i am able to access the variables outside the class as they are public 
variables,suppose outside the class i should not be able to provide it as 
an access
'''
```

```python
dir(person)
```

```python
#private variable using('__variable')

class Person:
    def __init__(self,name,age,gender):
        self.__name=name   #private variables
        self.__age=age     #private variables
        self.gender=gender #public variables

def get_name(person):
    return person.__name

person= Person("yash",22,"male")
get_name(person) #not able to directly access private variable
'''
it cannot be used outside the class it can only be acessed inside the class 
we have to use getter and setter methods '''
```

```python
person=Person("Yash",22,"male")
dir(person)
#You will see only public variables are visible at bottom
```

```python
#protected variables

class Person:
    def __init__(self,name,age,gender):
        self._name=name   #protected variables(single underscore)
        self._age=age  
        self.gender=gender

class Employee(Person):
    def __init__ (self ,name,age,gender):
        super().__init__(name,age,gender) #dont use self keyword here

employee = Employee("Yash",22,"Male")
print(employee._name)
'''
you cannot use it outside the class but we can access it with a derived class

'''
```

```python
## Encapsulatiokn with getter and setter method 

class Person:
    def __init__(self,name,age):
        self.__name=name 
        self.__age=age

        #getter method
    def get_name(self):
        return self.__name #called from inside the class 
    
        #setter method (helps to change any value)
    def set_name(self,name):
        self.__name=name

    #getter method for age
    def get_age(self):
        return self.__age
    
    #setter method 
    def set_age(self,age):
        if age>0:
            self.__age=age
        else:
            print("Age cannot be negative")

person= Person("Yash",22)

#acess modify private variables using getter and setter method 
print(person.get_name())
print(person.get_age())


person.set_age(23)
print(person.get_age())

person.set_age(-5)
```

### Abstraction
***
abstraction is the concept of hiding the complex implementation details and showing only the necessary feature of an object this helps in reducing programming complexity and effort

```python
'''
everyone uses washing machine ,buttons are visible,but internal process are hidden
'''
from abc import ABC,abstractmethod

#Abstract base class

class Vehicle(ABC):
    def drive(self):
        print("the vehicle is used for driving ")

    @abstractmethod
    def start_engine(self):
        pass #go ahead and implement the function in your own way

class Car(Vehicle):
    def start_engine(self):
        print("car engine started ")

def operate_vehicle(vehicle):
    vehicle.start_engine()

car=Car()
operate_vehicle(car)
''' there is one functionality that is available for you,defined in form of abstract method
if any child class wants to inherit , you have to make sure you have to implement the function 
in your own way (the knife can be used to cut,spread butter whatever you want it to do) '''
```

### Magic Methods

Magic methods in python, also known as dunder methods (double underscore methods ),are special methods that start and end with double underscores .these methods enable you to define the beavior of objects for built in operations such as arithemetic operations and comparisons and more

### Magic Methods

Magic methods are predefined methods in python that you can overide to change the behaviour of your objects .some common magic methods include

* __init__:initializzes a new instance of class
* str: returns a string representatin of an obj
* repr: return an official string representation of an obj
* len: returns the length of an object
* getitem: gets an items from container
* setitem ; sets an item in a  container


```python
#basic methods 
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):  #overiding the magic method 
        return f"{self.name},{self.age} years old"
    
    def __repr__(self):#again overiding
        return f"Person(name={self.name},age={self.age})"
person = Person("yash",22)
print(person)
print(repr(person))
```

### Operator Overloading 
***
operator overloading allows you to define the behaviour of operators (+,-,*,/) for custom objects you achieve thi s by overriding specific magic methods in your class

```python
### common operator overloading 

'''
__add__(self,other): Adds two objects using the  + operator
__sub__(self,other): Subtracts two objects using the - operator
__mul__(self,other): mul
__truediv__(self,other):divide
__eq_(self,other):checks if two objects are equal
__lt__(self,other): checks if one object is less tha other obj
'''
```

```python
#custom mathematical operation on vectors
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    
    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    
    def __mul__(self,other):
        return Vector(self.x*other,self.y*other)
    
    def __repr__(self):
        return f'Vector ({self.x,self.y})'
    
v1=Vector(2,3)
v2=Vector(4,5)

print(v1+v2)
```

### Practical Example :Reading Large files

Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory

```python
### practical example :Reading large files
def read_largefile(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_python = "largefile.txt"
for line in read_largefile(file_path):
    print(line.strip())
```

### decorators 

Decorators are a powerful and flexible featur in python that allows you to modify the behaviour of a function or class method . theya re commonly used to add funcitonalty to functions or methods without modifing their actual code .

```python
#function copy
#closures 
#decorators
```

```python
##closures functions process of execution
def main_welcome():
    msg='welcome'
    def sub_welcome_method():
        print('welcome to the advance python course')
        print(msg)#can be access in the sub method if defined outside the sub methode
        print("learn these concepts properly")

    return sub_welcome_method()
```

```python
main_welcome()
```

```python
def main_welcome(func):
  
    def sub_welcome_method():
        print('welcome to the advance python course')
        func('welcome everyone to ths tutorial')
        print("learn these concepts properly")

    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
### decorator
def main_welcome(func):
    
    def sub_welcome_method():
        print('welcome to the advance python course')
        func()
        print("learn these concepts properly")

    return sub_welcome_method()
```

```python
def course_introduction():
    print('this is an advance python course ')

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print('this is an advance python course ')
'''
when we are using this decorator the function course_introduction is going inside the 
decorator and triggering the function in which the course_introduction wll act as parameter
WITHOUT MODIFYING THE ACTUAL FUNCTION
'''
```
