### Python memory management

memory management in python involves a combination of automatic garbage collection,reference counting,and various internal optimization to efficiently manage memory allocatrion and deallocation. understanding these mechanisms can help developers write more efficient and robust applications

### reference counting 

refernce counting is the primary methodf python uses to manage memoryt rach object in python maintains a count of references pointing to it when the reference count drops to zero the memory occupied by the object is deallocated.

```python
import sys

a=[]
##2 (one ref from 'a', and one from  getrefcount)
print(sys.getrefcount(a))
```

```python
b=a
print(sys.getrefcount(b))
```

```python
del b 
print(sys.getrefcount(a))
#when the count becomes '0' the memory gets deallocated
```

### Garbage Collection
***

python includes cyclic garbage collector to handle references cycles.References cycles occur when objects reference each other, preventing their refrences counts from reaching zero

```python
import gc
#enable garbage collection
gc.enable()
```

```python
gc.disable()
```

```python
gc.collect()
#shows the value of unreachable objects in the system
```

```python
##get garbage collection stats
print(gc.get_stats())
```

```python
## get unreachable objects
print(gc.garbage)
```

#### Memory management best practices
***
* use local variable:local variable have shorte lifespan and are freed sooner than global variables
* Avoid circular references: Circular references can lead to memory leaks if not properly managed. 
* Used Generator: Generators produce items one at a time only keep one item in memory at a time, making them memory efficient
* Explicitly delete Objects: use the del statement to delete variables and objects explicitly.
* Profile Memory Usage:Use memory profiling tools like tracemalloc and memory profiler to identify memory leaks nad optimize memory usage.

```python
#handling circular reference
import gc 

class myobject:
    def __init__(self,name):
        self.name=name
        print(f'Objects{self.name} created')

    def __del__(self):
        print(f"object {self.name} deleted")#its better to delete when object is not in use


#create circular references
obj1=myobject("obj1")
obj2=myobject('obj2')

obj1.ref=obj2
obj2.ref=obj1 #created circular references which is a bad thing as the best practices suggest to avoid circular references

del obj1
del obj2

#manually trigger the garbage collection
#have to do it because of circular reference
gc.collect()
```

```python
#print colelcted objects
print(f'Garbage collected objects:{gc.garbage}')
```

```python
#generators for memory efficiency
#generators allow u to produce an item at a time,using memory efficiently by only keeping one item 

def generate_numbers(n):
    for i in range(n):
        yield i 

#using the generator

for num in generate_numbers(1000000):
    print(num)
    if num>10:
        break
```

```python
##Profiling memory usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats= snapshot.statistics('lineno')

    print('[Top 10]')
    for stat in top_stats[:10]:#top 10 results
        print(stat)
```

```python
main()
```
